```python
def predict_riasec_from_answers(answers):
    """Predict RIASEC personality type from user answers"""
    text = " ".join(answers).lower()
    vec = vectorizer.transform([text])
    return riasec_model.predict(vec)[0]

print("Prediction function ready")
```

```python
def Interest_questionnaire():
    """The RIASEC interest questionnaire"""
    
    answers = []
    
    print("=" * 60)
    print("RIASEC INTEREST QUESTIONNAIRE")
    print("=" * 60)
    print("Please type your answer to each question (be detailed)\n")
    
    questions = [
        "Do you prefer working with your hands, tools, or machines, or do you prefer working with ideas and data?",
        "When you have free time, do you prefer building or fixing things, reading and research, or creating art or music?",
        "Are you more drawn to solving logical problems (math, puzzles) or creative problems (designing, brainstorming)?",
        "Do you enjoy leading groups, organizing events, or persuading others?",
        "Do you prefer helping people directly (teaching, nursing, counseling) or working with data/things?",
        "Do you like having clear rules and routines, or do you prefer flexibility and variety?",
        "Is it important to you that others see you as an expert or leader in your field?",
        "Would you prefer a career where you are well-known and respected, or one where you work behind the scenes?",
        "Do you want your work to be recognized with awards, titles, or public acknowledgment?",
        "How important is it that your work directly helps other people or improves society?",
        "Would you sacrifice a higher salary to work for an organization that makes a positive social impact?",
        "Do you feel motivated by work that addresses problems like poverty, disease, or environmental issues?",
        "Do you prefer working alone, in small teams, or with large groups of people?",
        "Are you comfortable with frequent meetings, presentations, and group discussions?",
        "Do you enjoy networking, building relationships, and communicating with many different people?",
        "Would you describe yourself as someone who listens carefully to others' problems and concerns?"
    ]
    
    for i, q in enumerate(questions, 1):
        print(f"\n{i}. {q}")
        ans = input("   Your answer: ").lower() 
        print(f"Recorded")
    
    return answers
```

```python
print("\n" + "=" * 60)
print("STARTING ASSESSMENT")
print("=" * 60)

# Get user input
student_answers = Interest_questionnaire()

# Display results
print("\n" + "=" * 60)
print("YOUR RESULTS")
print("=" * 60)

print("\nYour answers:")
for i, ans in enumerate(student_answers, 1):
    preview = ans[:50] + "..." if len(ans) > 50 else ans
    print(f"   {i}. {preview}")

# Predict personality
personality = predict_riasec_from_answers(student_answers)
print(f"\nDetected Personality Type: {personality.upper()}")

descriptions = {
    "Realistic": "You're a hands-on problem solver who enjoys working with tools, machines, and building things.",
    "Investigative": "You're an analytical thinker who enjoys research, data, and solving complex problems.",
    "Artistic": "You're a creative individual who values self-expression through art, design, or music.",
    "Social": "You're a people-oriented helper who enjoys teaching, counseling, and supporting others.",
    "Enterprising": "You're a persuasive leader who enjoys influencing others and managing projects.",
    "Conventional": "You're an organized individual who prefers structure, clear rules, and attention to detail."
}

print(f"\n{descriptions[personality]}")

# Show career recommendations
print(f"\nRecommended Career Fields for {personality}:")
for field in riasec_to_fields[personality]:
    print(f"{field}")

# Find specific subject requirements from database
print(f"\nRecommended Subjects to Focus On:")
for field in riasec_to_fields[personality]:
    print(f"\n   For {field}:")
    matching_subjects = [item for item in recommandation_database if item['field_name'] == field]
    for subject in matching_subjects[:3]:
        req_type = subject['requirement_type']
        print(f"      - {subject['subject_name']} ({req_type}): {subject['descriptions']}")

print("\n" + "=" * 60)
print("Assessment Complete!")
print("=" * 60)
```
